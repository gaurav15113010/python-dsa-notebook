two_sum([2, 7, 11, 15], 9)  # (0, 1)
two_sum([3, 2, 4], 6)       # (1, 2)
two_sum([3, 3], 6)          # (0, 1)
two_sum([1, 2, 3], 10)     # None

```python
def two_sum(nums,target):
    n = len(nums)
    ref_dict = {}
    for i in range(n):
        k = target - nums[i]
        if k in ref_dict:
            return (i,ref_dict[k])
        else:
            ref_dict[nums[i]]=i
```
