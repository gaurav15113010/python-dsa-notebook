```python
list1=[-1,2,3,2,2,9,5]
largest=float("-inf")
second_largest=float("-inf")
list1 = list(set(list1))
for i in list1:
    if i> largest:
        second_largest=largest
        largest=i
    elif largest>i>second_largest:
        second_largest=i

print(second_largest)
print(largest)
```

Armstrong number

```python
n=153
num=n
k=0  # can find this directly as well by : len(str(n))
sumn=0
while n>0:
    n=n//10
    k+=1

while num>0:
    ld = num%10
    sumn+=ld**k
    num=num//10

print(sumn)
```

Find Factorial of a Number

```python
def facto(n):
    if n==1:
        return 1
    return n * facto(n-1)
facto(5)
```

Reverse an array using recursion

```python
num=[5,7,3,2,6,1,5,9]
l=0
r=7
def reverse_fun(num,l,r):
    if l>=r:
        return num
    num[l],num[r] = num[r],num[l]
    return reverse_fun(num,l+1,r-1)

reverse_fun(num,l,r)
```

Reverseing an array using while loop

```python
num=[5,7,3,2,6,1,5,9]
l=0
r=7
while l<=r:
    num[l],num[r] = num[r],num[l]
    l+=1
    r-=1
print(num)



```

Check if a string is Palindrome using recursion/Loop

```python
s="mom"
l=0
r=len(s)-1
new_s = s
def palindrome(s,l,r):
    if l>=r:
        return "True Plaindrome"
    if s[l]==s[r]:
        return palindrome(s,l+1,r-1)
    else:
        print("Not Palidrome")
        return


    
palindrome(s,l,r)
```

Fibonacci Number using Recursion

```python
def fibo(n):
    if n==0 or n==1:
        return n
    return fibo(n-1)+fibo(n-2)

fibo(5)
```

Move Zeros to the end of the list

```python
nums=[1,2,0,4,3,0,0,3,5,1]
#expected o/p = [1,2,4,3,3,5,1,0,0,0]
i=0
while i<len(nums):
    if nums[i]!=0:
        i+=1
    else:
        break
j=i+1
while j<len(nums):
    if nums[j]!=0:
        nums[i],nums[j]=nums[j],nums[i]
        i+=1
    j+=1

nums   
```
