Find the Second Largest Number from a given list

```python
list1=[-1,2,3,2,2,9,5]
largest=float("-inf")
second_largest=float("-inf")
list1 = list(set(list1))
for i in list1:
    if i> largest:
        second_largest=largest
        largest=i
    elif largest>i>second_largest:
        second_largest=i

print(second_largest)
print(largest)
```

check palindrome in numbers

```python
n=1221
num=n
nn=0
while n>0:
    ld = n%10
    nn=nn*10+ld
    n=n//10
print(nn)
if nn==num:
    print("True")
```

Armstrong number

```python
n=153
num=n
k=0  # can find this directly as well by : len(str(n))
sumn=0
while n>0:
    n=n//10
    k+=1

while num>0:
    ld = num%10
    sumn+=ld**k
    num=num//10

print(sumn)
```

Hashing in Python.
Prestoring values into some data structure like (List/dict/sets) and fetching it

```python
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,67,2]
```

```python
hash_n = {}
for i in range(1,len(n)):
    hash_n[n[i]] = hash_n.get(n[i],0)+1
hash_n 
```

```python
for k in range(len(m)):
    print(hash_n.get(m[k],0))
```

Character Hashing

```python
s = "azyxyyzaaa"
q=["d","a","y","x"]
```

```python
hash_dict={}
for char in s:
    hash_dict[char] = hash_dict.get(char,0)+1
hash_dict    
```

```python
for i in q:
    print(hash_dict.get(i,0))
```

Recursion: Print X , N times, recursion using parameters

```python
def fun1(x,n):
    if n==0:
        return "finish"
    print(x)
    return fun1(x,n-1)

fun1('x',4)
```

Recursion: Sum of N natural Numbers

```python
def sum_func(add,i,n): #parametrized, just printing
    if i>n:
        print(add)
        return
    sum_func(add+i,i+1,n)

sum_func(0,1,4)
```

Functional Recursion : Create a flow, create base condition

Q: 1 to N sum

```python
def sum_func2(n):
    if n==1:
        return 1
    return n + sum_func2(n-1)
sum_func2(4)
```

Find Factorial of a Number

```python
def facto(n):
    if n==1:
        return 1
    return n * facto(n-1)
facto(5)
```
