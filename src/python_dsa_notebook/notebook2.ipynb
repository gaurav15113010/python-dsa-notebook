Find the Second Largest Number from a given list

```python
list1=[-1,2,3,2,2,9,5]
largest=float("-inf")
second_largest=float("-inf")
list1 = list(set(list1))
for i in list1:
    if i> largest:
        second_largest=largest
        largest=i
    elif largest>i>second_largest:
        second_largest=i

print(second_largest)
print(largest)
```

check palindrome in numbers

```python
n=1221
num=n
nn=0
while n>0:
    ld = n%10
    nn=nn*10+ld
    n=n//10
print(nn)
if nn==num:
    print("True")
```

Armstrong number

```python
n=153
num=n
k=0  # can find this directly as well by : len(str(n))
sumn=0
while n>0:
    n=n//10
    k+=1

while num>0:
    ld = num%10
    sumn+=ld**k
    num=num//10

print(sumn)
```
