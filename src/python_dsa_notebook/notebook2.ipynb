Find the Second Largest Number from a given list

```python
list1=[-1,2,3,2,2,9,5]
largest=float("-inf")
second_largest=float("-inf")
list1 = list(set(list1))
for i in list1:
    if i> largest:
        second_largest=largest
        largest=i
    elif largest>i>second_largest:
        second_largest=i

print(second_largest)
print(largest)
    
      
```
